# E-Commerce Sales Analytics Dashboard

**Project type:** Data analysis and visualization using Pandas and Matplotlib  
**Topic:** Online shopping store performance  
**Period:** January to June

This project uses a sample dataset to track orders, customers, new customers, products sold, website visits, revenue, and customer reviews. Run the cells from top to bottom.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("ecommerce_sales_1000_rows.csv", encoding= 'unicode_escape')
```

```python
# 2. BUILD THE DASHBOARD
fig, axes = plt.subplots(4, 2, figsize=(15, 17))
fig.suptitle(
    "E-Commerce Sales Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)

# Chart 1: Orders over time
df.plot(x="Month", y="Orders", ax=axes[0, 0],
        title="Monthly Orders", marker="o", color="royalblue")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Number of Orders")
axes[0, 0].grid(axis="y", alpha=0.3)

# Chart 2: Active customers
df.plot(x="Month", y="ActiveCustomers", kind="bar", ax=axes[0, 1],
        title="Active Customers", color="mediumseagreen")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Customers")
axes[0, 1].grid(axis="y", alpha=0.3)

# Chart 3: New customer growth
df.plot(x="Month", y="NewCustomers", kind="area", ax=axes[1, 0],
        title="New Customers", color="mediumorchid", alpha=0.6)
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("New Customers")
axes[1, 0].grid(axis="y", alpha=0.3)

# Chart 4: Products sold
df.plot(x="Month", y="ProductsSold", ax=axes[1, 1],
        title="Products Sold Trend", marker="o", color="darkorange")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Products Sold")
axes[1, 1].grid(axis="y", alpha=0.3)

# Chart 5: Revenue
df.plot(x="Month", y="Revenue", kind="bar", ax=axes[2, 0],
        title="Monthly Revenue", color="crimson")
axes[2, 0].set_xlabel("Month")
axes[2, 0].set_ylabel("Revenue ($)")
axes[2, 0].grid(axis="y", alpha=0.3)

# Chart 6: Customer reviews
df.plot(x="Month", y="Reviews", ax=axes[2, 1],
        title="Customer Reviews", marker="o", color="teal")
axes[2, 1].set_xlabel("Month")
axes[2, 1].set_ylabel("Number of Reviews")
axes[2, 1].grid(axis="y", alpha=0.3)

# Chart 7: Active vs returning customers
df.plot(x="Month", y=["ActiveCustomers", "ReturningCustomers"],
        kind="bar", ax=axes[3, 0],
        title="Active vs Returning Customers",
        color=["mediumseagreen", "mediumpurple"])
axes[3, 0].set_xlabel("Month")
axes[3, 0].set_ylabel("Customers")
axes[3, 0].grid(axis="y", alpha=0.3)

# Chart 8: Website visits vs orders
axes[3, 1].scatter(df["WebsiteVisits"], df["Orders"],
                   color="gold", edgecolors="black", s=100)
axes[3, 1].set_title("Website Visits vs Orders")
axes[3, 1].set_xlabel("Website Visits")
axes[3, 1].set_ylabel("Orders")
axes[3, 1].grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
# 3. BASIC DATA ANALYSIS
print("----- Basic Data Analysis -----")
print("Total Orders:", df["Orders"].sum())
print("Average Active Customers:", round(df["ActiveCustomers"].mean(), 2))
print("Total Products Sold:", df["ProductsSold"].sum())
print("Total Website Visits:", df["WebsiteVisits"].sum())
print("Total Revenue: $", f"{df['Revenue'].sum():,}")
print("Average Customer Reviews:", round(df["Reviews"].mean(), 2))

# Useful additional metric: average order value
average_order_value = df["Revenue"].sum() / df["Orders"].sum()
print("Average Order Value: $", round(average_order_value, 2))
```

```python
# 4. HIGHEST MONTHLY VALUES
print("\\n----- Highest Values -----")

for column, label, suffix in [
    ("Orders", "Orders", ""),
    ("ActiveCustomers", "Active Customers", ""),
    ("Revenue", "Revenue", "$"),
    ("ProductsSold", "Products Sold", ""),
    ("WebsiteVisits", "Website Visits", "")
]:
    idx = df[column].idxmax()
    print(f"{label}: {df.loc[idx, 'Month']} ({suffix}{df.loc[idx, column]:,})")
```

```python
# 5. KEY INSIGHTS
print("\\n----- Key Insights -----")
print("1. Monthly orders rise overall from January through June.")
print("2. Active customers grow across the six-month period.")
print("3. June has the highest number of new customers.")
print("4. Products sold generally increase alongside orders.")
print("5. Revenue is highest in June.")
print("6. Customer reviews rise overall, which may reflect greater order volume.")
print("7. Returning customers increase over the period.")
print("8. Website visits and orders can be compared to explore traffic-to-order patterns.")
print("\\nNote: This is illustrative sample data, not actual store records.")
```
